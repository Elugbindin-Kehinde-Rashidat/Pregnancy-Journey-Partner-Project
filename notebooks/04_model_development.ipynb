# Model Development
## Objectives

This notebook focuses on developing machine learning models for maternal health risk prediction.

The following tasks will be performed:

- Load the processed datasets.
- Split the data into training and testing sets.
- Train a Logistic Regression baseline model.
- Train Decision Tree and Random Forest models.
- Perform hyperparameter tuning on the Random Forest model.
- Save the trained models for evaluation and deployment.

```python
# Import Libraries

import polars as pl
from pathlib import Path

import joblib

from sklearn.model_selection import train_test_split
from sklearn.model_selection import GridSearchCV

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
# Display Settings

pl.Config.set_tbl_cols(-1)
pl.Config.set_tbl_rows(20)
```

## Load Standardized Dataset

Logistic Regression performs better when numerical features are standardized. Therefore, the standardized dataset created during preprocessing will be used as the baseline model.

```python
scaled_df = pl.read_csv(
    "../data/processed/maternal_health_standardized.csv"
)

scaled_df.head()
```

```python
# Features and Target

X = scaled_df.drop("risk_level").to_numpy()

y = scaled_df["risk_level"].to_numpy()
```

## Observation

The predictor variables (X) and target variable (y) have been successfully separated.

```python
X_train, X_test, y_train, y_test = train_test_split(

    X,

    y,

    test_size=0.20,

    random_state=42,

    stratify=y
)
```

```python
print("Training Samples:", X_train.shape)

print("Testing Samples:", X_test.shape)
```

## Observation

The dataset was divided into training and testing sets using an 80:20 ratio.

Stratified sampling was applied to preserve the distribution of maternal risk categories in both datasets.

# Baseline Model
## Logistic Regression

Logistic Regression serves as the baseline model because it is simple, computationally efficient, and widely used as a benchmark for classification problems.

```python
logistic_model = LogisticRegression(

    max_iter=1000,

    random_state=42
)

logistic_model.fit(

    X_train,

    y_train
)
```

```python
print("Baseline Logistic Regression trained successfully.")
```

## Observation

The baseline Logistic Regression model has been successfully trained using the standardized dataset.

# Decision Tree
## Load Non-Standardized Dataset

Decision Trees do not require feature scaling because they split data using decision rules rather than distances.

```python
processed_df = pl.read_csv(
    "../data/processed/maternal_health_processed.csv"
)

processed_df.head()
```

```python
X_tree = processed_df.drop("risk_level").to_numpy()

y_tree = processed_df["risk_level"].to_numpy()
```

```python
X_train_tree, X_test_tree, y_train_tree, y_test_tree = train_test_split(

    X_tree,

    y_tree,

    test_size=0.20,

    random_state=42,

    stratify=y_tree
)
```

```python
decision_tree = DecisionTreeClassifier(

    random_state=42
)

decision_tree.fit(

    X_train_tree,

    y_train_tree
)
```

```python
print("Decision Tree trained successfully.")
```

## Observation

The Decision Tree model was successfully trained using the processed dataset.

# Random Forest
Random Forest combines multiple decision trees to improve prediction accuracy and reduce overfitting.

```python
random_forest = RandomForestClassifier(

    random_state=42
)

random_forest.fit(

    X_train_tree,

    y_train_tree
)
```

```python
print("Random Forest trained successfully.")
```

## Observation

The Random Forest model has been trained using the default hyperparameters and will serve as the reference model before optimisation.

# Hyperparameter Tuning

Hyperparameter tuning searches for the combination of parameters that produces the best model performance.

```python
param_grid = {

    "n_estimators":[100,200,300],

    "max_depth":[10,20,None],

    "min_samples_split":[2,5,10],

    "min_samples_leaf":[1,2,4]
}
```

```python
grid_search = GridSearchCV(
    estimator=random_forest,
    param_grid=param_grid,
    cv=5,
    scoring="recall_macro",
    n_jobs=-1,
    verbose=2
)
```

```python
grid_search.fit(

    X_train_tree,

    y_train_tree
)
```

```python
best_random_forest = grid_search.best_estimator_

print(grid_search.best_params_)
```

## Observation

The Random Forest model was optimised using GridSearchCV. The best combination of hyperparameters was selected based on macro-average Recall.

# Save Models

```python
Path("../models").mkdir(

    exist_ok=True
)
```

```python
# Save Logistic Regression model
joblib.dump(
    logistic_model,
    "../models/logistic_regression.pkl"
)

# Save Decision Tree model
joblib.dump(
    decision_tree,
    "../models/decision_tree.pkl"
)

# Save Random Forest model
joblib.dump(
    random_forest,
    "../models/random_forest.pkl"
)

# Save Tuned Random Forest model
joblib.dump(
    best_random_forest,
    "../models/tuned_random_forest.pkl"
)

print("All models saved successfully!")
```

```python
import polars as pl
from pathlib import Path

Path("../data/processed").mkdir(
    exist_ok=True
)


# Logistic Regression test data

logistic_test = pl.DataFrame(
    X_test
)


logistic_test = logistic_test.with_columns(
    pl.Series(
        name="risk_level",
        values=y_test
    )
)


logistic_test.write_csv(
    "../data/processed/logistic_test_data.csv"
)


print("Logistic Regression test data saved!")
```

```python
# Decision Tree and Random Forest test data

tree_test = pl.DataFrame(
    X_test_tree
)


tree_test = tree_test.with_columns(
    pl.Series(
        name="risk_level",
        values=y_test_tree
    )
)


tree_test.write_csv(
    "../data/processed/tree_models_test_data.csv"
)


print("Tree models test data saved!")
```
