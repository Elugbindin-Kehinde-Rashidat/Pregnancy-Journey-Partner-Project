```python
# Import the Polars library
import polars as pl
```

```python
# Load the first dataset
df1 = pl.read_csv("../data/raw/maternal_high_burden.csv")

# Load the second dataset
df2 = pl.read_csv("../data/raw/maternal_low_burden.csv")

# Load the third dataset
df3 = pl.read_csv("../data/raw/maternal_moderate_burden.csv")
```

```python
print(df1.shape)
print(df2.shape)
print(df3.shape)
```

```python
# Combine all three datasets vertically
merged_df = pl.concat([df1, df2, df3])
```

```python
merged_df.shape
```

```python
merged_df.write_csv("../data/raw/maternal_health_dataset.csv")
```
